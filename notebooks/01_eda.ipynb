Checking failure information

```python
# 0--> for NO imminent engine failure 
#  1 --> Engine failure is imminent


print("Engine failures:")
print(df["engine_failure_imminent"].value_counts())

print("\nBrake failures:")
print(df["brake_issue_imminent"].value_counts())

print("\nBattery failures:")
print(df["battery_issue_imminent"].value_counts())

print("\nFailure types:")
print(df["failure_type"].value_counts())
```

```python
df["failure"] = (
    (df["engine_failure_imminent"] == 1) |
    (df["brake_issue_imminent"] == 1) |
    (df["battery_issue_imminent"] == 1)
).astype(int)

print(df["failure"].value_counts())

# failure = 0 → Normal
#  failure = 1 → Failure risk
```
